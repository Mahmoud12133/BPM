```python
# 1. Reload or read raw timestamp strings so we can reliably detect full date-time vs. date-only
# Replace with your actual filename (or use raw string column)

# 2. Filter out empty case_ids
valid_case = df['case_id'].notna() & (df['case_id'].astype(str).str.strip() != '')

# 3. Filter to rows with a FULL date-time timestamp (must have time component ':' and parse cleanly)
raw_ts = df['timestamp'].astype(str).str.strip()
has_time = df['timestamp'].notna() & raw_ts.str.contains(':')
parsed_ts = pd.to_datetime(raw_ts, errors='coerce')

# Keep only valid rows, preserving exact original FILE order (do NOT sort!)
sub = df[valid_case & has_time & parsed_ts.notna()].copy()
sub['ts'] = parsed_ts[valid_case & has_time & parsed_ts.notna()]

# 4. Check if timestamps are non-decreasing per case_id in file order
# is_monotonic_increasing allows equal consecutive timestamps (non-decreasing: t1 <= t2 <= t3)
is_non_decreasing = sub.groupby('case_id', sort=False)['ts'].is_monotonic_increasing

out_of_order_cases = sorted(is_non_decreasing[~is_non_decreasing].index.tolist())

print("Count:", len(out_of_order_cases))
print("Cases:", out_of_order_cases)
```

```python
import pandas as pd

# Identify missing or empty resource values
is_missing_res = df['resource'].isna() | (df['resource'].astype(str).str.strip() == '')

print(f"Total rows: {len(df)} | Missing resource rows: {is_missing_res.sum()} ({is_missing_res.mean():.1%})")

# 1. Check where missing values occur in file/row order (start, middle, end, or all through)
missing_indices = df.index[is_missing_res]
print(f"First missing row index: {missing_indices.min()}, Last missing row index: {missing_indices.max()}")

# 2. Check missing rate across time buckets (e.g., 10 equal chunks from start to end of file/timeline)
df['chunk'] = pd.qcut(df.index, q=10, labels=False)
print("\nMissing rate by chronological/file decile (0 = start -> 9 = end):")
print(is_missing_res.groupby(df['chunk']).mean().round(3))

# 3. Check missing rate by activity (to see if it only happens on specific activities)
print("\nMissing rate by normalized activity:")
act_norm = df['activity'].astype(str).str.replace('_', ' ').str.strip().str.lower()
print(is_missing_res.groupby(act_norm).agg(['count', 'sum', 'mean']).round(3))
```

```python
import pandas as pd

# Reload raw CSV to ensure uncleaned values are checked accurately
raw = pd.read_csv(r"C:\Users\moudy\Downloads\20237012\20237012\Q1_raw_event_log.csv")  # Replace with your filename

# Helper masks
empty_case = raw['case_id'].isna() | (raw['case_id'].astype(str).str.strip() == '')
raw_ts = raw['timestamp'].astype(str).str.strip()
empty_ts = raw['timestamp'].isna() | (raw_ts == '')
has_time = (~empty_ts) & raw_ts.str.contains(':')
parsed_ts = pd.to_datetime(raw_ts, errors='coerce')

# Normalize activity
act_norm = raw['activity'].astype(str).str.replace('_', ' ').str.strip().str.lower()

# 1. TS-MIS (Q1.4): Missing/empty timestamps
q1_4_ts_mis = empty_ts.sum()

# 2. TS-IMP (Q1.5): Imprecise timestamps (present, but missing time of day)
q1_5_ts_imp = ((~empty_ts) & (~has_time)).sum()

# 3. RES-MIS (Q1.6): Missing/empty resource
q1_6_res_mis = (raw['resource'].isna() | (raw['resource'].astype(str).str.strip() == '')).sum()

# 4. EV-CIL (Q1.7): Event-level colliding/duplicate rows (exact duplicates or same case+activity+ts)
q1_7_ev_cil = raw.duplicated().sum()

# 5. ACT-IMP (Q1.9): Imprecise/non-canonical activity labels (not already clean Title Case)
canonical_acts = {
    'File Claim', 'Assign Adjuster', 'Inspect Damage', 'Request Evidence',
    'Assess Liability', 'Approve Payout', 'Notify Claimant', 'Close Claim'
}
q1_9_act_imp = (~raw['activity'].isin(canonical_acts)).sum()

# 6. CASE-MIS (Q1.3): Missing/empty case_id
q1_3_case_mis = empty_case.sum()

# 7. EV-MIL (Q1.10): Non-empty cases missing mandatory activity ('request evidence')
valid_df = raw[~empty_case].copy()
valid_df['act_norm'] = act_norm[~empty_case]
all_cases = set(valid_df['case_id'].unique())
has_mand = set(valid_df.loc[valid_df['act_norm'] == 'request evidence', 'case_id'])
q1_10_ev_mil = len(all_cases - has_mand)

# 8. POS-INC (Q1.11): Non-empty cases with non-monotonic timestamps in file order (full datetime rows only)
sub = raw[(~empty_case) & has_time & parsed_ts.notna()].copy()
sub['ts'] = parsed_ts[(~empty_case) & has_time & parsed_ts.notna()]
mono = sub.groupby('case_id', sort=False)['ts'].is_monotonic_increasing
q1_11_pos_inc = (~mono).sum()

counts = {
    "1 = TS-MIS (Q1.4)": q1_4_ts_mis,
    "2 = TS-IMP (Q1.5)": q1_5_ts_imp,
    "3 = RES-MIS (Q1.6)": q1_6_res_mis,
    "4 = EV-CIL (Q1.7)": q1_7_ev_cil,
    "5 = ACT-IMP (Q1.9)": q1_9_act_imp,
    "6 = CASE-MIS (Q1.3)": q1_3_case_mis,
    "7 = EV-MIL (Q1.10)": q1_10_ev_mil,
    "8 = POS-INC (Q1.11)": q1_11_pos_inc,
}

for label, val in counts.items():
    print(f"{label:<22}: {val}")

winner = max(counts, key=counts.get)
print(f"\nHighest count -> {winner} ({counts[winner]})")
```
